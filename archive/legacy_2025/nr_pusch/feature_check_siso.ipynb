```python
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
import plotly.express as px

# Load the dataset
df = pd.read_csv("dataset_sionna_siso.csv")
df = df.reset_index()

# Select the features for visualization
features = df[["snr", "highest_received_power", "kurtosis", "skewness", "mean", "std", "delay_spread", "normalized_delay_power_correlation"]]

# Convert complex numbers to their absolute values
features["mean"] = features["mean"].apply(lambda x: abs(complex(x)))
features["normalized_delay_power_correlation"] = features["normalized_delay_power_correlation"].apply(lambda x: abs(complex(x)))

# Perform PCA
pca = PCA(n_components=2)
pca_result = pca.fit_transform(features)

# Create a DataFrame for PCA results
pca_df = pd.DataFrame(pca_result, columns=['PCA1', 'PCA2'])

# Plot each feature against the PCA components
for feature in features.columns:
    pca_df[feature] = features[feature].values
    fig = px.scatter(pca_df, x='PCA1', y='PCA2', color=feature, title=f'PCA of Features colored by {feature}')
    fig.show()

# Perform t-SNE
tsne = TSNE(n_components=2, random_state=42)
tsne_result = tsne.fit_transform(features)

# Create a DataFrame for t-SNE results
tsne_df = pd.DataFrame(tsne_result, columns=['TSNE1', 'TSNE2'])

# Plot each feature against the t-SNE components
for feature in features.columns:
    tsne_df[feature] = features[feature].values
    fig = px.scatter(tsne_df, x='TSNE1', y='TSNE2', color=feature, title=f't-SNE of Features colored by {feature}')
    fig.show()
```

```python
import pandas as pd
from sklearn.manifold import TSNE
import plotly.express as px

# Load the dataset
df = pd.read_csv("dataset_sionna_siso.csv")
df = df.reset_index()

# Select the features for visualization
features = df[["snr", "highest_received_power", "kurtosis", "skewness", "mean", "std", "delay_spread", "normalized_delay_power_correlation"]]

# Convert complex numbers to their absolute values
features["mean"] = features["mean"].apply(lambda x: abs(complex(x)))
features["normalized_delay_power_correlation"] = features["normalized_delay_power_correlation"].apply(lambda x: abs(complex(x)))

# Transpose the DataFrame to have features as rows
feature_data = features.transpose()

# Add a column for feature names
feature_data['feature'] = feature_data.index
print(feature_data.head)
# Perform t-SNE with a lower perplexity value
tsne = TSNE(n_components=2, random_state=42, perplexity=5)
tsne_result = tsne.fit_transform(feature_data.drop(columns=['feature']))

# Create a DataFrame for t-SNE results
tsne_df = pd.DataFrame(tsne_result, columns=['TSNE1', 'TSNE2'])
tsne_df['feature'] = feature_data['feature'].values

# Plot t-SNE results
fig_tsne = px.scatter(tsne_df, x='TSNE1', y='TSNE2', color='feature', title='t-SNE of Features')
fig_tsne.show()
```
